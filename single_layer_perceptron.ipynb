```python
from typing import List, Callable, Dict, TypedDict

import random
```

```python
def step(weighted_sum: float, threshold: float = 0.0) -> int:
    if weighted_sum > threshold:
        return 1

    else:
        return 0
```

```python
class Perceptron:
    def __init__(self, weights: List[float], bias: float, activation_function: Callable[[float], float]):
        self.weights = weights
        self.bias = bias
        self.activation_function = activation_function

    def predict(self, inputs: List[float]):
        # Check if if there is a weight for each input
        if len(self.weights) != len(inputs):
            raise ValueError("Number of inputs must match the number of weights.")

        weighted_sum: float = 0.0

        # Calculate the weighted sum
        for i in range(len(inputs)):
            product = inputs[i] * self.weights[i]
            weighted_sum += product

        weighted_sum += self.bias

        output = self.activation_function(weighted_sum)

        return output
```

```python
class Movie(TypedDict):
    genre: str
    duration: float
    actor: str
```

```python
class Encoder:
    def __init__(self) -> None:
        self.genres = ["Action", "Comedy", "Thriller"]
        self.max_duration = 300.00 # Minutes 

    def preprocess(self, movie: Movie):
        inputs: List[float] = []

        genre = movie.get("genre")

        # One-Hot Encoding
        for g in self.genres:
            if g == genre:
                inputs.append(1.0)

            else:
                inputs.append(0.0)

        duration = movie.get("duration")

        # Standardise the input feature for 'duration'
        duration = duration / self.max_duration

        inputs.append(duration)

        if movie.get("actor") == "The Rock":
            inputs.append(1.0)

        else:
            inputs.append(0.0)

        return inputs
```

```python
movie: Movie = {
    "genre": "Action",
    "duration": 120,
    "actor": "The Rock"
}

encoder = Encoder()

inputs = encoder.preprocess(movie)

weights = [random.uniform(-0.5, 0.5) for _ in range(len(inputs))]

bias = -0.5

perceptron = Perceptron(weights, bias, step)

output = perceptron.predict(inputs)
print(output)
```
