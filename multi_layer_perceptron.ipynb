```python
from typing import List, Callable

import random
import math
```

Activation Function

```python
def relu(weighted_sum: float) -> float:
    return max(0.0, weighted_sum)

def sigmoid(weighted_sum: float):
    return 1 / (1 + math.exp(-weighted_sum))
```

Neuron

```python
class Neuron:
    def __init__(self, input_dims: int) -> None:
        self.weights = [random.uniform(-0.5, 0.5) for _ in range(input_dims)]
        self.bias = 0

    # Calculate the weighted sum of a neuron
    def forward(self, inputs: List[float]) -> float:
        if len(inputs) != len(self.weights):
            raise ValueError("The number of inputs must match the number of weights")

        weighted_sum = 0.0

        for i in range(len(inputs)):
            product = inputs[i] * self.weights[i]
            weighted_sum += product

        weighted_sum += self.bias

        return weighted_sum
```

Hidden Layer

```python
class DenseLayer:
    def __init__(self, input_dims: int, output_dims: int, activation_fn: Callable[[float], float]) -> None:
        self.neurons = [Neuron(input_dims) for _ in range(output_dims)]
        self.activation_fn = activation_fn

    # Calculate the outputs of each neuron in the given layer, this will be the inputs to the neuron in the next layer
    def forward(self, inputs: List[float]) -> List[float]:
        outputs: List[float] = []

        for neuron in self.neurons:
            z = neuron.forward(inputs)
            a = self.activation_fn(z)
            
            outputs.append(a)

        return outputs
```

```python
class MultiLayerPerceptron:
    def __init__(self, layer_dims: List[int]) -> None:
        if len(layer_dims) < 2:
            raise ValueError("Neural Network must have at least input and output layers.")
        
        self.layers: List[DenseLayer] = []

        # Create the layer
        for i in range(len(layer_dims) - 1):
            is_last_layer = i == len(layer_dims) - 2

            if is_last_layer:
                activation_fn = sigmoid
            else:
                activation_fn = relu

            layer = DenseLayer(layer_dims[i], layer_dims[i + 1], activation_fn)
            
            self.layers.append(layer)

    def forward(self, inputs: List[float]):
        new_inputs = inputs

        for layer in self.layers:
            new_inputs = layer.forward(new_inputs)

        return new_inputs       
```

```python
mlp = MultiLayerPerceptron([2,3,1])

ingredients = [0.9, 0.1]

mlp.forward(ingredients)
```
