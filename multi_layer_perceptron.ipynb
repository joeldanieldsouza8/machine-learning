Activation Function

```python
def relu(weighted_sum: float) -> float:
    return max(0.0, weighted_sum)
```

Neuron

```python
class Neuron:
    def __init__(self, input_dim: int) -> None:
        self.weights = [random.uniform(-0.5, 0.5) for _ in range(input_dim)]
        self.bias = random.uniform(-0.5, 0.5)

    def forward(self, inputs: List[float]) -> float:
        if len(inputs) != len(self.weights):
            raise ValueError("The number of inputs must match the number of weights")

        weighted_sum = 0.0

        for i in range(len(inputs)):
            product = inputs[i] * self.weights[i]
            weighted_sum += product

        weighted_sum += self.bias

        return weighted_sum
```

Hidden Layer

```python
class DenseLayer:
    def __init__(self, input_dim: int, units: int, activation: Callable[[float], float]) -> None:
        self.neurons = [Neuron(input_dim) for _ in range(units)]
        self.activation = activation

    def forward(self, inputs: List[float]) -> List[float]:
        outputs: List[float] = []

        for neuron in self.neurons:
            outputs.append(self.activation(neuron.forward(inputs)))

        return outputs
```
